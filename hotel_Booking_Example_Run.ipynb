```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    classification_report,
    confusion_matrix
)

def evaluate_model(model, X, y, threshold=0.5):

    probabilities = model.predict_proba(X)[:, 1]

    predictions = (
        probabilities >= threshold
    ).astype(int)

    print(f"Accuracy:  {accuracy_score(y, predictions):.4f}")
    print(f"Precision: {precision_score(y, predictions, zero_division=0):.4f}")
    print(f"Recall:    {recall_score(y, predictions, zero_division=0):.4f}")
    print(f"F1 Score:  {f1_score(y, predictions, zero_division=0):.4f}")
    print(f"ROC-AUC:   {roc_auc_score(y, probabilities):.4f}")
    print(f"PR-AUC:    {average_precision_score(y, probabilities):.4f}")

    print("\nConfusion Matrix:")
    print(confusion_matrix(y, predictions))

    print("\nClassification Report:")
    print(classification_report(y, predictions, zero_division=0))

    return probabilities, predictions
```

```python
def engineer_features(df):

    df = df.copy()

    # Total number of guests
    df['total_guests'] = (
        df['adults'].fillna(0) +
        df['children'].fillna(0) +
        df['babies'].fillna(0)
    )

    # Total stay duration
    df['total_nights'] = (
        df['stays_in_weekend_nights'].fillna(0) +
        df['stays_in_week_nights'].fillna(0)
    )

    # Previous booking history
    df['previous_booking_history'] = (
        df['previous_cancellations'].fillna(0) +
        df['previous_bookings_not_canceled'].fillna(0)
    )

    # Total special requests
    df['has_special_request'] = (
        df['total_of_special_requests'].fillna(0) > 0
    ).astype(int)

    return df
```

```python
df_city_hotel_train = engineer_features(df_city_hotel_train)
df_city_hotel_val = engineer_features(df_city_hotel_val)
df_city_hotel_test = engineer_features(df_city_hotel_test)
```

```python
LEAKY_AND_UNUSED_COLS = [
    'is_canceled',
    'reservation_status',
    'reservation_status_date',
    'arrival_date',
    'hotel'
]

def prepare_xy(df):

    X = df.drop(
        columns=LEAKY_AND_UNUSED_COLS,
        errors='ignore'
    ).copy()

    y = df['is_canceled'].copy()

    # Convert agent/company into binary indicators
    if 'agent' in X.columns:
        X['has_agent'] = X['agent'].notna().astype(int)
        X.drop(columns='agent', inplace=True)

    if 'company' in X.columns:
        X['has_company'] = X['company'].notna().astype(int)
        X.drop(columns='company', inplace=True)

    return X, y


X_train, y_train = prepare_xy(df_city_hotel_train)
X_val, y_val = prepare_xy(df_city_hotel_val)
X_test, y_test = prepare_xy(df_city_hotel_test)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

categorical_cols = X_train.select_dtypes(
    include=['object', 'category', 'string']
).columns.tolist()

numeric_cols = X_train.select_dtypes(
    include=[np.number, 'bool']
).columns.tolist()

numeric_transformer = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ]
)

categorical_transformer = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('onehot', OneHotEncoder(
            handle_unknown='ignore',
            sparse_output=True
        ))
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_cols),
        ('cat', categorical_transformer, categorical_cols)
    ]
)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

C_values = [0.001, 0.01, 0.1, 1.0, 10.0, 100.0]

best_c = None
best_val_auc = -np.inf

for c_val in C_values:

    clf = Pipeline(
        steps=[
            ('preprocessor', preprocessor),
            ('classifier', LogisticRegression(
                C=c_val,
                max_iter=2000,
                random_state=42
            ))
        ]
    )

    clf.fit(X_train, y_train)

    val_probabilities = clf.predict_proba(X_val)[:, 1]

    val_auc = roc_auc_score(
        y_val,
        val_probabilities
    )

    print(
        f'C={c_val:<8} | Validation ROC-AUC: {val_auc:.4f}'
    )

    if val_auc > best_val_auc:
        best_val_auc = val_auc
        best_c = c_val

print('\nBest C:', best_c)
print('Best Validation ROC-AUC:', round(best_val_auc, 4))
```

```python
X_train_full = pd.concat(
    [X_train, X_val],
    axis=0
)

y_train_full = pd.concat(
    [y_train, y_val],
    axis=0
)

final_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('classifier', LogisticRegression(
            C=best_c,
            max_iter=2000,
            random_state=42
        ))
    ]
)

final_pipeline.fit(
    X_train_full,
    y_train_full
)
```

```python
def monthly_cancellation_analysis(df):

    data = df.copy()

    data['arrival_month'] = (
        data['arrival_date'].dt.to_period('M').astype(str)
    )

    monthly = (
        data.groupby('arrival_month')
        .agg(
            total_bookings=('is_canceled', 'count'),
            canceled_bookings=('is_canceled', 'sum'),
            cancellation_rate=('is_canceled', 'mean')
        )
        .reset_index()
    )

    fig, ax1 = plt.subplots(figsize=(15, 5))

    sns.barplot(
        data=monthly,
        x='arrival_month',
        y='total_bookings',
        color='lightblue',
        ax=ax1
    )

    ax1.set_ylabel('Total Bookings')

    ax2 = ax1.twinx()

    sns.lineplot(
        data=monthly,
        x='arrival_month',
        y='cancellation_rate',
        marker='o',
        color='red',
        ax=ax2
    )

    ax2.set_ylabel('Cancellation Rate')

    plt.title('Monthly Booking Volume and Cancellation Rate')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

    return monthly


monthly_stats = monthly_cancellation_analysis(
    df_city_hotel_train
)

print(monthly_stats)
```

```python
categorical_features = [
    'meal',
    'market_segment',
    'distribution_channel',
    'reserved_room_type',
    'assigned_room_type',
    'deposit_type',
    'customer_type',
    'arrival_date_month',
    'is_repeated_guest',
    'previous_cancellations',
    'required_car_parking_spaces'
]

categorical_features = [
    col for col in categorical_features
    if col in df_city_hotel_train.columns
]


def analyze_categorical_features(df, features):

    results = {}

    for col in features:

        stats = (
            df.groupby(col, dropna=False)
            .agg(
                total_bookings=('is_canceled', 'size'),
                canceled_bookings=('is_canceled', 'sum'),
                cancellation_rate=('is_canceled', 'mean')
            )
            .reset_index()
        )

        stats['booking_percentage'] = (
            stats['total_bookings'] / len(df) * 100
        )

        stats = stats.sort_values(
            'cancellation_rate',
            ascending=False
        )

        results[col] = stats

        print(f"\n{'='*60}")
        print(f"Feature: {col}")
        print('='*60)

        print(stats.to_string(index=False))

    return results


categorical_results = analyze_categorical_features(
    df_city_hotel_train,
    categorical_features
)
```

```python
def create_revenue_features(df):

    df = df.copy()

    # Total guests
    df['total_guests'] = (
        df['adults'].fillna(0) +
        df['children'].fillna(0) +
        df['babies'].fillna(0)
    )

    # Total stay duration
    df['total_stays_in_nights'] = (
        df['stays_in_weekend_nights'].fillna(0) +
        df['stays_in_week_nights'].fillna(0)
    )

    # Guest history
    df['previous_booking_history'] = (
        df['previous_cancellations'].fillna(0) +
        df['previous_bookings_not_canceled'].fillna(0)
    )

    # Agent/company presence
    if 'agent' in df.columns:
        df['has_agent'] = df['agent'].notna().astype(int)

    if 'company' in df.columns:
        df['has_company'] = df['company'].notna().astype(int)

    # Arrival date features
    if 'arrival_date' in df.columns:

        arrival = pd.to_datetime(df['arrival_date'])

        df['arrival_month_num'] = arrival.dt.month
        df['arrival_day_num'] = arrival.dt.day
        df['arrival_day_of_week'] = arrival.dt.dayofweek

    return df
```

```python
EXCLUDED_COLS = [
    'is_canceled',
    'reservation_status',
    'reservation_status_date',
    'arrival_date',
    'arrival_date_year',
    'arrival_date_month',
    'arrival_date_week_number',
    'arrival_date_day_of_month',
    'agent',
    'company',
    'booking_changes',
    'assigned_room_type',
    'country'
]

# Note:
# We are extracting date features separately.
# Country is excluded initially to avoid high-cardinality
# complexity in this first baseline.
# It can be tested in a subsequent experiment.

def prepare_revenue_xy(df):

    data = create_revenue_features(df)

    X = data.drop(
        columns=EXCLUDED_COLS,
        errors='ignore'
    ).copy()

    y = data['is_canceled'].copy()

    return X, y
```

```python
def simulate_cancellations(
    probabilities,
    simulations=20000,
    random_state=42
):

    rng = np.random.default_rng(random_state)

    probabilities = np.asarray(probabilities)

    simulated = rng.binomial(
        n=1,
        p=probabilities,
        size=(simulations, len(probabilities))
    )

    return simulated.sum(axis=1)
```

```python
def optimize_overselling(
    existing_probabilities,
    room_capacity,
    room_contribution,
    walking_cost,
    new_booking_cancel_prob,
    max_extra_bookings=30,
    simulations=20000,
    random_state=42
):

    rng = np.random.default_rng(random_state)

    existing_probabilities = np.asarray(
        existing_probabilities,
        dtype=float
    )

    if (
        np.any(existing_probabilities < 0) or
        np.any(existing_probabilities > 1)
    ):
        raise ValueError("Probabilities must be between 0 and 1.")

    if not 0 <= new_booking_cancel_prob <= 1:
        raise ValueError("New booking probability must be between 0 and 1.")

    results = []

    # Simulate cancellations among existing bookings
    existing_cancellations = rng.binomial(
        1,
        existing_probabilities,
        size=(simulations, len(existing_probabilities))
    ).sum(axis=1)

    for extra_bookings in range(max_extra_bookings + 1):

        if extra_bookings > 0:

            new_cancellations = rng.binomial(
                1,
                new_booking_cancel_prob,
                size=(simulations, extra_bookings)
            ).sum(axis=1)

        else:
            new_cancellations = np.zeros(simulations)

        total_bookings = (
            len(existing_probabilities) + extra_bookings
        )

        total_cancellations = (
            existing_cancellations + new_cancellations
        )

        actual_arrivals = (
            total_bookings - total_cancellations
        )

        occupied_rooms = np.minimum(
            actual_arrivals,
            room_capacity
        )

        walked_guests = np.maximum(
            actual_arrivals - room_capacity,
            0
        )

        revenue = (
            occupied_rooms * room_contribution
        )

        relocation_cost = (
            walked_guests * walking_cost
        )

        profit = revenue - relocation_cost

        results.append({
            'Extra_Bookings': extra_bookings,
            'Total_Bookings': total_bookings,
            'Expected_Cancellations': total_cancellations.mean(),
            'Expected_Occupancy': occupied_rooms.mean(),
            'Expected_Walked_Guests': walked_guests.mean(),
            'Probability_Of_Walking': (
                walked_guests > 0
            ).mean(),
            'Expected_Revenue': revenue.mean(),
            'Expected_Walking_Cost': relocation_cost.mean(),
            'Expected_Profit': profit.mean()
        })

    results_df = pd.DataFrame(results)

    optimal_row = results_df.loc[
        results_df['Expected_Profit'].idxmax()
    ]

    return results_df, optimal_row
```

```python
example_results, optimal_decision = optimize_overselling(
    existing_probabilities=np.array([
        0.80, 0.65, 0.40, 0.20, 0.10
    ]),
    room_capacity=4,
    room_contribution=120,
    walking_cost=250,
    new_booking_cancel_prob=0.30,
    max_extra_bookings=8,
    simulations=20000
)

print("Optimal decision:")
print(optimal_decision)

display(example_results)
```

```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler

X_train = df_city_hotel_train.drop(columns=['is_canceled']),
y_train = df_city_hotel_train['is_canceled']
X_val = df_city_hotel_val.drop(columns=['is_canceled']),
y_val = df_city_hotel_val['is_canceled'],
X_test = df_city_hotel_test.drop(columns=['is_canceled']),
y_test = df_city_hotel_test['is_canceled']

# 1. Fit the scaler ONLY on the training set, then transform all three
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

# 2. Tune hyperparameters using the Validation set
best_c = None
best_val_acc = 0.0

for c_val in [0.001, 0.01, 0.1, 1.0, 10.0, 100.0]:
    model = LogisticRegression(C=c_val, max_iter=1000, random_state=42)
    model.fit(X_train_scaled, y_train)

    val_preds = model.predict(X_val_scaled)
    val_acc = accuracy_score(y_val, val_preds)
    print(f"C={c_val:<6} | Validation Accuracy: {val_acc:.4f}")

    if val_acc > best_val_acc:
        best_val_acc = val_acc
        best_c = c_val

print(f"\nBest C selected: {best_c} (Val Accuracy: {best_val_acc:.4f})")

# 3. Retrain the final model on combined Train + Validation data
X_train_full = np.vstack((X_train, X_val))
y_train_full = np.concatenate((y_train, y_val))

final_scaler = StandardScaler()
X_train_full_scaled = final_scaler.fit_transform(X_train_full)
X_test_final_scaled = final_scaler.transform(X_test)

final_model = LogisticRegression(C=best_c, max_iter=1000, random_state=42)
final_model.fit(X_train_full_scaled, y_train_full)

# 4. Evaluate once on the unseen Test set
test_preds = final_model.predict(X_test_final_scaled)
test_probs = final_model.predict_proba(X_test_final_scaled)[:, 1]

print("\n--- Final Test Set Results ---")
print(f"Test Accuracy: {accuracy_score(y_test, test_preds):.4f}")
print("\nConfusion Matrix:\n", confusion_matrix(y_test, test_preds))
print("\nClassification Report:\n", classification_report(y_test, test_preds))
```
